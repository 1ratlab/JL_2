#  ***Full Ingestion Pipeline for IB770 + IB970*** 

# Implementation ***************************************

***def Block 1 — Load raw tables from Excel***

```python
import pandas as pd

def load_ib770(path):
    return pd.read_excel(path)

def load_ib970(path):
    return pd.read_excel(path)
```

***🔧 Block 2 — Apply the IB770 → IB970 mapping dictionary***

```python
label_map = {
    "LBM of Right Arm": "Lean Mass of Right Arm",
    "LBM% of Right Arm": "Lean Mass(%) of Right Arm",
    "LBM of Left Arm": "Lean Mass of Left Arm",
    "LBM% of Left Arm": "Lean Mass(%) of Left Arm",
    "LBM of Trunk": "Lean Mass of Trunk",
    "LBM% of Trunk": "Lean Mass(%) of Trunk",
    "LBM of Right Leg": "Lean Mass of Right Leg",
    "LBM% of Right Leg": "Lean Mass(%) of Right Leg",
    "LBM of Left Leg": "Lean Mass of Left Leg",
    "LBM% of Left Leg": "Lean Mass(%) of Left Leg",
    "TBW/LBM": "TBW/FFM",
    "ICW/LBM%": "ICW/FFM%",
    "LBM Control": "FFM Control",
    # All other columns map to themselves
}
```

***Make into def To make this dictionary complete and robust:***

```python
def complete_mapping(label_map, ib770_cols, ib970_cols):
    full_map = {}

    for col in ib970_cols:
        if col in label_map:
            full_map[col] = label_map[col]
        elif col in ib770_cols:
            full_map[col] = col
        else:
            full_map[col] = None  # IB970‑unique → blank column
    return full_map
```

***📐 Block 3 — Normalize any input (770 or 970) into the IB970 schema***

```python
def harmonize(df_raw, ib970_columns, full_map):
    df_out = pd.DataFrame(columns=ib970_columns)

    for ib97_col in ib970_columns:
        ib77_col = full_map.get(ib97_col)

        if ib77_col is None:
            df_out[ib97_col] = pd.NA
        elif ib77_col in df_raw.columns:
            df_out[ib97_col] = df_raw[ib77_col]
        else:
            df_out[ib97_col] = pd.NA

    return df_out
```

***🧬 Block 4 — Unified ingestion function***
*This is the operator‑grade pipeline you will call from your notebook:*

```python
def ingest_inbody(path, device_type, ib970_columns, label_map):
    df_raw = pd.read_excel(path)

    # Build complete mapping dictionary
    full_map = complete_mapping(label_map, df_raw.columns, ib970_columns)

    # Normalize into IB970 schema
    df_norm = harmonize(df_raw, ib970_columns, full_map)

    # Add device metadata
    df_norm["source_device"] = device_type
    df_norm["source_file"] = path

    return df_norm
```

***Usage:***

```python
df_770 = ingest_inbody("ib770.xlsx", "IB770", ib970_columns, label_map)
df_970 = ingest_inbody("ib970.xlsx", "IB970", ib970_columns, label_map)
```

***Then unify:***

```python
df_all = pd.concat([df_770, df_970], ignore_index=True)
```
