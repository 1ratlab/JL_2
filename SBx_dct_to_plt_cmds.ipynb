```python
def import_table(sheet_name):
    import pandas as pd
    print("Edits are made to the 'onedrive' version of 'dat_mnl.xlsm")
    print("HAS LATEST onedrive version of 'dat_mnl.xlsm' BEEN COPIED TO THE REPO DIR?? __ wsl.localhost_Ubuntu-20.04_home_ratlabs_JL_2_data_dat_mnl_dat_mnl.xlsm")
    from openpyxl import load_workbook
    import warnings
    warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")  

    
    xl_path = "/home/ratlabs/JL_2/data/dat_mnl/dat_mnl.xlsm"
    wb = load_workbook(xl_path, data_only=True)

    ws = wb[sheet_name]

    if not ws.tables:
        raise ValueError(f"No tables found in sheet '{sheet_name}'")

    # get first table
    tbl_name = list(ws.tables.keys())[0]
    tbl = ws.tables[tbl_name]
    ref = tbl.ref  # e.g. "A1:F20"

    # --- FIX: convert "A1:F20" → "A:F" ---
    start_cell, end_cell = ref.split(":")
    start_col = ''.join(filter(str.isalpha, start_cell))
    end_col   = ''.join(filter(str.isalpha, end_cell))
    col_range = f"{start_col}:{end_col}"

    df = pd.read_excel(
        xl_path,
        sheet_name=sheet_name,
        engine="openpyxl",
        usecols=col_range
    )

    return df
```

```python
def get_ctrl_attrs(unique_name, instruction_dct):
    """
    Returns the list of attribute values for the row whose key-0 matches unique_name.
    Output list excludes the unique_name itself.
    """

    for row_key, row_dict in instruction_dct.items():
        if row_dict.get(0) == unique_name:
            # Extract keys 1..N in sorted order
            return [row_dict[k] for k in sorted(row_dict.keys()) if k != 0]

    raise KeyError(f"Control name '{unique_name}' not found.")
```
