## ***defs*** --- Functions* 

```python
def load_folder_of_CSVfiles_to_df_raw(folder_path, filename, pattern="*"):
    """
    ALL CSV FILES MUST HAVE THE SAME NUMBER OF DATA COLS 
    and A HEADER WITH THE SAME COL LABLES
    Multiple rows may work but should be tested first
    -----
    It Load data using 'universal_import'  with explicit folder_path and filename.
    It Imports all csv files in the folder
    one file for each row of data
    when all files have been opened and contents added to a data array
    It creates a dataframe df_raw from the data and as a return
    """
    full_path = f"{folder_path}/{filename}"
    df_raw = universal_import(full_path, pattern=pattern)
    return df_raw
```

## Final Harmonization Function (Single Unified Dictionary)

```python
def harmonize_inbody(df_raw, unified_map):
    """
    Harmonize IB770 or IB970 data into the unified nickname namespace.

    Steps:
    1. Rename columns using unified_map.
    2. Report unmapped columns.
    3. Return normalized DataFrame.
    """

    # 1. Rename columns
    df_norm = df_raw.copy()
    df_norm.columns = [
        unified_map.get(col, col)   # fallback: keep original if unmapped
        for col in df_norm.columns
    ]

    # 2. Report unmapped columns
    original_cols = set(df_raw.columns)
    mapped_cols = set(unified_map.keys())
    unmapped = sorted(list(original_cols - mapped_cols))

    if unmapped:
        print("\n⚠️ Unmapped columns detected:")
        for col in unmapped:
            print(f"   - {col}")
    else:
        print("\n✓ All columns successfully mapped.")

    # 3. Return normalized DataFrame
    return df_norm
```
