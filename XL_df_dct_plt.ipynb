## def functions

```python
import pandas as pd
from openpyxl import load_workbook

def import_plt_style_table():
    """
    Import the Excel TABLE named 'plt_styl' from dat_mnl.xlsm
    and return it as df_plt_styl_tbl.
    """

    xl_path = "/home/ratlabs/JL_2/data/dat_mnl/dat_mnl.xlsm"

    # Load workbook in NORMAL mode (not read-only)
    wb = load_workbook(xl_path, data_only=True)

    # Find the sheet containing the table
    for ws in wb.worksheets:
        if "plt_styl" in ws.tables:
            tbl = ws.tables["plt_styl"]
            sheet_name = ws.title
            ref = tbl.ref  # e.g., "A1:F20"
            break
    else:
        raise ValueError("Table 'plt_styl' not found in workbook.")

    # Import ONLY the table range
    df_plt_styl_tbl = pd.read_excel(
        xl_path,
        sheet_name=sheet_name,
        engine="openpyxl",
        usecols=ref
    )

    return df_plt_styl_tbl
```

```python
def build_plt_styl_dct():
    """
    Convert a style table (df_plt_styl_tbl) into a dictionary (plt_styl_dct)
    where:
        - Outer key = row index
        - Inner key 0 = style name (plt_styl_nm)
        - Inner keys 1..N = attribute values from the row
    """
    plt_styl_dct = {}

    for idx, row in df_plt_styl_tbl.iterrows():
        inner = {}

        # key 0 = style name
        inner[0] = row["plt_styl_nm"]

        # keys > 0 = remaining attributes
        col_keys = [c for c in df_plt_styl_tbl.columns if c != "plt_styl_nm"]

        for k, col in enumerate(col_keys, start=1):
            inner[k] = row[col]

        plt_styl_dct[idx] = inner

    return plt_styl_dct
```
